## Create an IAM User
You need credentials that allow Python to communicate with AWS.

    1. Open AWS IAM Console.
    2. Navigate to Users → Create user.
    3. Enter a username, for example rekognition-python-user.
    4. Grant the user the necessary Rekognition permissions.
    5. Create an access key for programmatic access.

For this exercise, the relevant permissions are:
Table:

IAM permission	| Purpose  |
|---|---:|
rekognition:DetectLabels	| Object detection |
rekognition:DetectText	| Text detection | 
rekognition:DetectModerationLabels	| Content moderation | 

For a learning environment, you can use the AWS-managed AmazonRekognitionFullAccess policy, although a custom policy with only the three permissions above is preferable.

AWS documentation: Rekognition IAM permissions.

Important: Keep your access key and secret access key private. Never upload them to GitHub or include them in your Python source code.
